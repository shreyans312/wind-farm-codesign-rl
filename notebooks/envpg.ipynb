# EnvPG

Ablaincourt layout and yaw co-design. Set the site and spacing limits before training.

```python
from pathlib import Path
import sys

ROOT = Path.cwd().resolve()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
assert (ROOT / "envpg.py").exists()
sys.path.insert(0, str(ROOT))
from envpg import Config, train
```

```python
SITE = None  # (xmin, xmax, ymin, ymax) in metres
MAX_DISPLACEMENT = None  # metres from each reference position
MIN_SPACING = None  # metres between turbines
RUN_NAME = "envpg-001"

assert SITE is not None and MAX_DISPLACEMENT is not None and MIN_SPACING is not None
config = Config(
    site=tuple(SITE),
    max_displacement=MAX_DISPLACEMENT,
    min_spacing=MIN_SPACING,
    run_dir=str(ROOT / "runs" / RUN_NAME),
)
train(config)
```
