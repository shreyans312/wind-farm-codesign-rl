# EnvPG

Ablaincourt layout and yaw co-design with the generalized critic. Site and spacing limits are chosen for this run.


```python
from pathlib import Path
import sys

ROOT = Path.cwd().resolve()
assert (ROOT / "envpg.py").exists()
sys.path.insert(0, str(ROOT))
from envpg import Config, train
```

```python
SITE = (400.0, 2500.0, -400.0, 400.0)
MAX_DISPLACEMENT = 150.0
MIN_SPACING = 250.0
RUN_NAME = "envpg-generalized-001"

config = Config(
    site=tuple(SITE),
    max_displacement=MAX_DISPLACEMENT,
    min_spacing=MIN_SPACING,
    run_dir=str(ROOT / "runs" / RUN_NAME),
)
train(config)
```
